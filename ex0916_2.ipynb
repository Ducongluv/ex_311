```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. 데이터 준비
# ============================================================

# seaborn에서 제공하는 tips 데이터셋을 불러온다.
tips = sns.load_dataset("tips")

# 데이터 확인
print(tips.head())


# ============================================================
# 2. 전체 Figure 생성
# ============================================================

# 전체 그래프를 담을 큰 도화지(Figure)를 생성한다.
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. 그래프 공간(Axes) 생성
# ============================================================

# 1행 1열 중 첫 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 히스토그램 그리기
# ============================================================

# total_bill 열의 데이터를 이용하여 히스토그램을 그린다.
#
# hist()
# → 숫자 데이터를 여러 구간(bin)으로 나누고
#   각 구간에 데이터가 몇 개 있는지 막대로 표현한다.
#
# bins=10
# → 전체 데이터를 10개의 구간으로 나눈다.
axes1.hist(tips["total_bill"], bins=10)


# ============================================================
# 5. 그래프 설정
# ============================================================

# 그래프 제목
axes1.set_title("Total Bill Distribution")

# x축 이름
axes1.set_xlabel("Total Bill")

# y축 이름
axes1.set_ylabel("Frequency")


# ============================================================
# 6. 그래프 간격 조정
# ============================================================

# 제목과 축 이름 등이 겹치지 않도록
# 그래프의 여백을 자동으로 조정한다.
fig.tight_layout()


# ============================================================
# 7. Figure 출력
# ============================================================

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. 데이터 준비
# ============================================================

# seaborn에서 제공하는 tips 데이터셋을 불러온다.
tips = sns.load_dataset("tips")

# 데이터 앞부분 확인
print(tips.head())


# ============================================================
# 2. 전체 Figure 생성
# ============================================================

# 전체 그래프를 담을 큰 도화지(Figure)를 생성한다.
fig = plt.figure(figsize=(8, 5))


# ============================================================
# 3. 그래프 공간(Axes) 생성
# ============================================================

# 1행 1열 중 첫 번째 위치에
# 그래프 공간(Axes)을 생성한다.
axes1 = fig.add_subplot(1, 1, 1)


# ============================================================
# 4. 산점도 그리기
# ============================================================

# x축 → 식사금액(total_bill)
# y축 → 팁(tip)
#
# 각 점 하나가 손님 한 명의 데이터를 나타낸다.
#
# 산점도는 두 숫자형 변수 사이의
# 관계나 패턴을 확인할 때 사용한다.
axes1.scatter(
    tips["total_bill"],
    tips["tip"]
)


# ============================================================
# 5. 그래프 설정
# ============================================================

# 그래프 제목
axes1.set_title("Total Bill vs Tip")

# x축 이름
axes1.set_xlabel("Total Bill")

# y축 이름
axes1.set_ylabel("Tip")


# ============================================================
# 6. 그래프 간격 조정
# ============================================================

# 제목과 축 이름 등이 잘 보이도록
# 그래프의 여백을 자동으로 조정한다.
fig.tight_layout()


# ============================================================
# 7. Figure 출력
# ============================================================

plt.show()
```
