```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
        "이름" : ["민수", "지영"],
        "점수" : [90, 85]
})

display(df1)
```

```python
df2 = pd.DataFrame([
    ["민수", 90],
    ["지영", 85],
], index = ["첫번째", "두번째"], columns= ["이름", "점수"])

df2
```

```python
df3 = pd.DataFrame([
    {"이름" : "민수", "점수": 90},
    {"이름" : "지영", "저음수": 100},

])
df3
```

```python
import numpy as np

df4 = pd.DataFrame(np.array([
    [1, 2], [3, 4]
]), columns = ["x", "y"])

df4
```

```python
# 이름, 나이, 성별 데이터로 DataFrame 생성
df = pd.DataFrame({
    '이름': ['철수', '영희', '민수', '지영', '준호'],
    '나이': [20, 21, 20, 22, 21],
    '성별': ['남', '여', '남', '여', '남']
})

print(df)
```
