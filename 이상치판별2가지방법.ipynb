```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
```

```python
arr = np.array([60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200])
print(arr)
print(type(arr))
```

```python
s = pd.Series([60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200])
print(s)
print(type(s))
```

```python
# IQR = Q3 - Q1
# 중앙값 = Q2

Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)
median = s.median()
print(Q1, median, Q3)
IQR = Q3 - Q1
print(IQR)
```

```python
upper = Q3 + (IQR*1.5)
lower = Q1 - (IQR*1.5)
print("upper : ", upper)
print("lower : ", lower)
```

```python
# 시각화 -> 박스플롯
fig = plt.figure()
axe1 = fig.add_subplot(1, 1, 1)
sns.boxplot(s)
plt.show()
```

```python
# Z-scroe : 평균에서 얼마나 멀리있는가?
mean = s.mean()
print(mean)

var = round(arr.var(), 1)
std = round(arr.std(), 1)
print(var, std)
```

```python
Z_score = (s - mean) / std
print(Z_score)
print(Z_score.abs() > 3)
```

```python
fig = plt.figure()
axe1 = fig.add_subplot(1, 1, 1)
sns.histplot(s, bins = "5")
plt.show()
```
